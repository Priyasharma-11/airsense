## Answers
* **Highest PM2.5 months:** December > November > January (winter inversions).
* **Consistently high locations:** Bhumkar Nagar, Thergaon, Savta Mali Nagar (Pimpri-Chinchwad belt).
* **Time of day:** morning peak ~08:00 IST, minimum ~15:00 IST.
* **Rainfall:** wet days are far cleaner (negative Spearman ρ).
* **Pollutant relationships:** PM2.5–PM10 strongest; NO₂ next (traffic).
* **Seasonality:** winter ≈ 4× monsoon.
